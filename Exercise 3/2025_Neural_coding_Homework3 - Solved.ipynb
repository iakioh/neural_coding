**(a)**

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson, norm, nbinom
```

```python
lambdas = [1, 5, 10, 50]

k_max = max(lambdas) * 2
k = np.arange(0, int(k_max) + 1)

plt.figure(figsize=(10, 6))
for lam in lambdas:
    pmf = poisson.pmf(k, mu=lam)
    plt.plot(k, pmf, linestyle='-', label=f'λ = {lam}')

plt.title('Poisson Probability Mass Functions')
plt.xlabel('k (number of events)')
plt.ylabel('P(X = k)')
plt.legend()
plt.grid(True)
plt.show()
```

**(b)**

Intuitively, it makes sense to use a Normal Distribution with mean $\mu = \lambda$ and variance $\sigma^2 = \lambda$. 

```python
fig, axes = plt.subplots(1, len(lambdas), figsize=(16, 4), sharey=True)

for ax, lam in zip(axes, lambdas):
    k_max = int(lam + 4 * np.sqrt(lam))
    k = np.arange(0, k_max + 1)
    
    pmf = poisson.pmf(k, mu=lam)
    ax.plot(k, pmf, marker='o', linestyle='-', label='Poisson PMF')
    
    x = np.linspace(0, k_max, 200)
    normal_pdf = norm.pdf(x, loc=lam, scale=np.sqrt(lam))
    ax.plot(x, normal_pdf, linestyle='--', label='Normal Approx.')
    
    ax.set_title(f'λ = {lam}')
    ax.set_xlabel('k')
    if lam == lambdas[0]:
        ax.set_ylabel('Probability')
    ax.grid(True)
    ax.legend()

plt.tight_layout()
plt.show()
```

We can see that as $\lambda$ increases, the normal approximation fits the poisson distribution better.


#### 1. Additive Gaussian Noise
- **Support violation**: Can produce negative firing rates, which are biologically implausible. Truncating or rectifying alters the mean and variance in a stimulus-dependent way.
- **Constant variance**: Variance is fixed at 1 regardless of firing rate, which does not match real neural data where variability typically increases with the mean (Fano factor ≥ 1).

#### 2. Multiplicative Gaussian Noise
- **Support violation**: Also allows negative values unless rectified, which distorts the distribution.
- **Overestimates variability**: Variance scales as $f(s)^2$, which grows faster than typical neural variability. Most neurons exhibit approximately linear scaling of variance with mean, not quadratic.

#### 3. Poisson Noise
- **Discrete values**: Models spike counts (integers), not continuous rates. Requires normalization if interpreting as rate.
- **Fixed mean-variance relation**: Enforces $\mathrm{Var}[r] = \mathbb{E}[r]$, which underestimates variability in overdispersed real neurons.
- **No temporal structure**: Assumes memoryless spiking (exponential interspike intervals), which neglects refractory periods and other dependencies.

#### 4. Negative Binomial Noise
- **Discrete values**: Like Poisson, outputs integer spike counts.
- **Flexible overdispersion**: Allows $\mathrm{Var}[r] > \mathbb{E}[r]$ via dispersion parameter $q$, better matching empirical Fano factors.
- **Still memoryless**: Captures overdispersion but not spike history effects like refractory periods.
